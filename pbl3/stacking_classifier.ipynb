# CODD PBL3 - Stacking Classifier (SVM + CART + GAM + Random Forest + MoE)
Uses the shared grouped splits and the same 30 features as every other model. Base models come from their notebooks (`saved_models/`). The meta-learner is trained on out-of-fold predictions from folds grouped by `Image_ID`.

```python
import joblib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

from codd_common import CLASS_NAMES, assert_feature_set, load_grouped_splits
from stacking_classifer import (MODEL_DIR, RESULTS_DIR, fit_stack, load_base_models, summarize)

data = load_grouped_splits()
for X in (data.X_train, data.X_val, data.X_test):
    assert_feature_set(X.columns)
base_models = load_base_models()
```

```python
# Validation: stack fit on the training split only
stack_val = fit_stack(base_models, data.X_train, data.y_train, data.groups_train)
val_pred = stack_val.predict(data.X_val)
val_row = summarize(data.y_val, val_pred, "Stacking Classifier", "Validation")
print(classification_report(data.y_val, val_pred, labels=CLASS_NAMES, zero_division=0))
```

```python
# Final model: refit on train + validation, evaluate once on the test split
X_tv = pd.concat([data.X_train, data.X_val], ignore_index=True)
y_tv = pd.concat([data.y_train, data.y_val], ignore_index=True)
g_tv = pd.concat([data.groups_train, data.groups_val], ignore_index=True)

final_stack = fit_stack(base_models, X_tv, y_tv, g_tv)
test_pred = final_stack.predict(data.X_test)
test_row = summarize(data.y_test, test_pred, "Stacking Classifier", "Test")
print(classification_report(data.y_test, test_pred, labels=CLASS_NAMES, zero_division=0))
metrics = pd.DataFrame([val_row, test_row])
metrics
```

```python
cm = confusion_matrix(data.y_test, test_pred, labels=CLASS_NAMES)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES)
plt.title("Stacking Classifier - Test Confusion Matrix")
plt.xlabel("Predicted Class"); plt.ylabel("Actual Class")
plt.tight_layout(); plt.show()
```

```python
metrics.to_csv(RESULTS_DIR / "stacking_metrics.csv", index=False)
joblib.dump(final_stack, MODEL_DIR / "codd_stacking_top5.joblib")
print("Saved stacking model and metrics.")
```
