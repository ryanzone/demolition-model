```python
import pandas as pd

train = pd.read_csv("../training_image_features.csv")
val = pd.read_csv("../validation_image_features.csv")
test = pd.read_csv("../testing_image_features.csv")

for col in ["Sample_ID", "Image_ID"]:
    if col in train.columns:
        train_ids = set(train[col].dropna())
        val_ids = set(val[col].dropna())
        test_ids = set(test[col].dropna())

        print(f"\n{col} overlap:")
        print("Train-Val:", len(train_ids & val_ids))
        print("Train-Test:", len(train_ids & test_ids))
        print("Val-Test:", len(val_ids & test_ids))
```

```python
target = "Condition_Class"

suspicious_columns = [
    "Condition_Score",
    "Damage_Score",
    "Structural_Integrity",
    "Recovery_Pathway",
    "Reconditioning_Required",
    "Recovery_Potential",
    "Alternative_Use_Category",
    "Alternative_Use_Suitability"
]

print("Target distribution:")
print(train[target].value_counts())

print("\nPotentially leaking features:")
for col in suspicious_columns:
    if col in train.columns:
        print(f"{col}: {train[col].nunique()} unique values")
```
